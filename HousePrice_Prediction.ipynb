# Detecting Outliers

```python
# Detecting outliers using IQR

import pandas as pd
import numpy as np

def iqr_outlier_summary(df, columns=None):
    if columns is None:
        columns = df.select_dtypes(include=[np.number]).columns.tolist()
    
    summary = []
    
    for col in columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
        
        summary.append({
            'Column': col,
            'Q1 (25%)': round(Q1, 4),
            'Q3 (75%)': round(Q3, 4),
            'IQR': round(IQR, 4),
            'Lower Bound': round(lower_bound, 4),
            'Upper Bound': round(upper_bound, 4),
            'Number of Outliers': len(outliers),
            'Percentage': f"{len(outliers)/len(df)*100:.2f}%",
          #  'vals': outliers[col].tolist()[:5] 
        })
    
    return pd.DataFrame(summary)


result = iqr_outlier_summary(df, df.columns)

print("="*100)
print("="*100)
print(result.to_string(index=False))
```

# Handle Outliers

```python
# Let's handle outliers in all columns 

def cap_outliers_multi(df, columns):
    for col in columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        df.loc[df[col] > upper_bound, col] = upper_bound
        df.loc[df[col] < lower_bound, col] = lower_bound
            
    return df

cols = ['CRIM', 'ZN', 'PTRATIO', 'B', 'LSTAT', 'RM', 'DIS']
df = cap_outliers_multi(df, cols)
```

# Fill Null values

```python
# I will fill the null values using KNNImputer from skylearn
from sklearn.impute import KNNImputer

def impute():
    features = ['CRIM', 'ZN', 'INDUS', 'CHAS', 'NOX', 'RM', 'AGE', 'DIS', 'RAD', 'TAX', 'PTRATIO', 'B', 'LSTAT']
    df_before = df.copy()
    imputer = KNNImputer(n_neighbors=5)
    
    df_imputed = imputer.fit_transform(df[features])
    
    df_imputed = pd.DataFrame(df_imputed, columns=features)
    
    df['AGE'] = df_imputed['AGE']
    df['CRIM'] = df_imputed['CRIM']
    df['ZN'] = df_imputed['ZN']
    df['INDUS'] = df_imputed['INDUS']
    df['CHAS'] = df_imputed['CHAS']
    df['LSTAT'] = df_imputed['LSTAT']
    
    return df
```

```python
df = impute()
```

# Data Split

```python
from sklearn.model_selection import train_test_split

target = df['MEDV']
cols = df.drop('MEDV', axis = 1)
X_train, X_test, y_train, y_test = train_test_split(cols, target, test_size = 0.2, random_state = 42)
```

# Creating polynomial features from existing columns

```python
from sklearn.preprocessing import PolynomialFeatures, StandardScaler,OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.feature_selection import SelectKBest, f_regression
from sklearn.linear_model import Ridge
```

```python
nonlinear_cols = ['LSTAT', 'RM', 'CRIM', 'AGE', 'NOX', 'INDUS', 'DIS']
linear_cols = ['ZN', 'TAX', 'PTRATIO', 'B']
discrete_cols = ['RAD'] 
binary_cols = ['CHAS']  

preprocessor = ColumnTransformer([
    ('poly_part', Pipeline([
        ('scaler', StandardScaler()),
        ('poly', PolynomialFeatures(degree=2, include_bias=False))
    ]), nonlinear_cols), 
    ('linear_part', StandardScaler(), linear_cols), 
    ('discrete_part', OneHotEncoder(drop='first', sparse_output=False), discrete_cols),
    ('binary_part', 'passthrough', binary_cols)
])

model = Pipeline([
    ('preprocessor', preprocessor),
    ('selector', SelectKBest(score_func=f_regression, k = 30)), 
    ('regressor', Ridge(alpha=1.0)) # Usign ridge to prevent model from overfitting
])
```

# Model Training

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

# Model Evaluation

```python
from sklearn.metrics import r2_score

print(f"R² Score: {r2_score(y_test, y_pred)}")
print(f"RMSE: {np.sqrt(np.mean((y_test - y_pred) ** 2))}")
```
